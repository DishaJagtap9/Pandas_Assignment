```python
## NP43. Scale dataset using NumPy functions.
#> 💡 **Hint:** Apply Min-Max or Z-score scaling using 
# NumPy on the `.values` array. Store results as new 
# columns in the DataFrame.
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Marks": [10, 20, 30, 40, 50]
})

arr = df["Marks"].values

# Min-Max Scaling
df["MinMax_Scaled"] = (arr - arr.min()) / (arr.max() - arr.min())

# Z-Score Scaling
df["Z_Score"] = (arr - arr.mean()) / arr.std()

print(df)
```

```python
### NP44. Engineer new features using NumPy.
#> 💡 **Hint:** Create polynomial features (`np.square`,
#  `np.power`), log features (`np.log1p`), interaction 
# features (column1 × column2), and binned features
#  (`np.digitize`).
import numpy as np
import pandas as pd

data = {
    "Name": ["Amit", "Riya", "John"],
    "Marks": [85, 90, 78],
    "Age": [22, 23, 24]
}

df = pd.DataFrame(data)

# Polynomial features
df["Square"] = np.square(df["Marks"])
df["Cube"] = np.power(df["Marks"], 3)

# Log feature
df["Log_Marks"] = np.log1p(df["Marks"])

# Interaction feature
df["Marks_Age"] = df["Marks"] * df["Age"]

# Binned feature
bins = [80, 90]

df["Binned"] = np.digitize(df["Marks"], bins)

print(df)
```

```python
## NP45. Use exponential functions (`np.exp`).
#> 💡 **Hint:** `np.exp(df["col"])` computes `
# e^x` for each element. 
# This is the inverse of `np.log`.
#  Useful for reverting log-transformed values.
import numpy as np
import pandas as pd

data = {
    "Name": ["Amit", "Riya", "John", "Sara", "Karan"],
    "Marks": [1, 2, 3, 4, 5]
}

df = pd.DataFrame(data)

# Apply exponential function
df["Exp_Marks"] = np.exp(df["Marks"])

print(df)

# Log transform
log_arr = np.log(df["Marks"])

# Revert back to original values
original_arr = np.exp(log_arr)

print("\nOriginal:")
print(df["Marks"])

print("\nLog:")
print(log_arr)

print("\nReverted:")
print(original_arr)
```

```python
### NP47. Apply conditional encoding using NumPy.
#> 💡 **Hint:** Use `np.select` with a list of 
# conditions and corresponding integer codes to encode 
# categorical columns numerically.
#  For example: `DS → 0`, `ML → 1`, `AI → 2`.
import numpy as np
import pandas as pd

data = {
    "Name": ["Amit", "Riya", "John", "Sara", "Karan"],
    "Course": ["DS", "ML", "AI", "DS", "ML"]
}

df = pd.DataFrame(data)

conditions = [
    df["Course"] == "DS",
    df["Course"] == "ML",
    df["Course"] == "AI"
]

choices = [0, 1, 2]

df["Course_Code"] = np.select(conditions, choices, default=-1)

print(df)

```

```python
### NP49. Perform a mini data analysis project.
#> 💡 **Hint:** Follow these steps: (1) Generate data
#with NumPy, (2) Store in DataFrame, 
#(3) Clean with Pandas, (4) Engineer features with NumPy, 
#(5) GroupBy analysis, (6) Detect outliers, (7) 
#Export summary.
import numpy as np
import pandas as pd

# Step 1: Generate data with NumPy
np.random.seed(42)

data = {
    "Department": np.random.choice(["IT", "HR", "Sales"], 20),
    "Salary": np.random.randint(20000, 100000, 20)
}

# Step 2: Store in DataFrame
df = pd.DataFrame(data)

# Step 3: Clean with Pandas
df.loc[3, "Salary"] = np.nan
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Step 4: Feature Engineering with NumPy
df["Bonus"] = np.round(df["Salary"] * 0.10, 2)
df["Salary_Squared"] = np.square(df["Salary"])

# Step 5: GroupBy Analysis
summary = df.groupby("Department")["Salary"].agg(
    ["count", "mean", "max", "min"]
)

# Step 6: Detect Outliers (IQR Method)
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

outliers = df[
    (df["Salary"] < Q1 - 1.5 * IQR) |
    (df["Salary"] > Q3 + 1.5 * IQR)
]

# Step 7: Export Summary
summary.to_csv("department_summary.csv")

print("DataFrame:")
print(df.head())

print("\nDepartment Summary:")
print(summary)

print("\nOutliers:")
print(outliers)



```

```python
### NP50. Build an end-to-end pipeline using NumPy + Pandas.
#> 💡 **Hint:** Chain all steps together:  
#> 1. `np.random.seed(X)` → generate raw data  

#> 2. `np.clip` → clean invalid values  
#> 3. `np.log`, `np.square`, `np.where` → feature engineering  
#> 4. `df.groupby().agg(...)` → statistical analysis  
#> 5. IQR + `np.clip` → outlier removal  
#> 6. `df[features].to_numpy(dtype=np.float32)` → ML feature matrix  
#> 7. Export with `df.to_csv(...)`
import numpy as np
import pandas as pd

# Step 1: Generate raw data
np.random.seed(45)
sales = np.random.randint(10, 100, 10)

# Step 2: Clean data
sales = np.clip(sales, 20, 80)

# Step 3: Feature Engineering
log_sales = np.log(sales)
square_sales = np.square(sales)
target = np.where(sales >= 50, 1, 0)

# Step 4: Create DataFrame
df = pd.DataFrame({
    "Sales": sales,
    "Log_Sales": log_sales,
    "Square_Sales": square_sales,
    "Target": target
})

# Add category feature
df["Category"] = np.where(df["Sales"] > 50, "High", "Low")

# Step 5: GroupBy Analysis
summary = df.groupby("Category")["Sales"].agg(
    ["count", "mean", "max", "min"]
)

print("Summary:")
print(summary)

# Step 6: IQR Outlier Removal
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Sales"] = np.clip(df["Sales"], lower, upper)

# Step 7: ML Feature Matrix
features = ["Sales", "Log_Sales", "Square_Sales"]

X = df[features].to_numpy(dtype=np.float32)
y = df["Target"].to_numpy()

print("\nX shape:", X.shape)
print("y shape:", y.shape)

# Step 8: Export
df.to_csv("final_pipeline.csv", index=False)
```
