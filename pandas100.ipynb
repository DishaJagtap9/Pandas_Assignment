```python
### Q4. Perform arithmetic operations on a Series.
#> 💡 **Hint:** Use `+`, `-`, `*`, `/`, `//`, `%`, `**` directly on the Series. Operations are **element-wise** and **index-aligned**.
import pandas as pd
s1=pd.Series([2,34,56,67,8])
s2=pd.Series([23,45,67,8,78])
#arithmatic operation Element wise 
print("addition of series:",s1+s2)
print("Substraction of series:",s1-s2)
print("Devision of series",s1/s2)
print("Floor Division",s1//s2)
print("square of series",s1**s2)
print("module of series",s1%s2)
#arithmatic operation as index wise
print("arithmatic operation as index wise")
i1=pd.Series([2,34,56,67,8],index=["a","b","c","d","e"])
i2=pd.Series([23,45,67,8,78],index=["b","a","c","d","e"])
print("addition of series:",i1+i2)
print("Substraction of series:",i1-i2)
print("Devision of series",i1/i2)
print("Floor Division",i1//i2)
print("square of series",i1**i2)
print("module of series",i1%2)

```

```python
### Q5. Find mean, min, max of a Series.
#> 💡 **Hint:** Use `.mean()`, `.min()`, `.max()`, `.sum()`, `.median()`, `.std()` directly on the Series.
s1=pd.Series([23,34,45,56,67,89])
print("mean of series",s1.mean())
print("minimum value of series",s1.min())
print("maximum value of series",s1.max())
print("total value of series:",s1.sum())
print("Median od series:")
print("standard deviation of series:",s1.std())
```

```python
### Q6. Apply conditional filtering on a Series.
#> 💡 **Hint:** Use a boolean mask: `s[s > 50]`. You can also chain conditions with `&` and `|
s1=pd.Series([34,45,67,78])
print(s1[s1>50])#single condition
print(s1[(s1>35) & (s1<50)])# double condition
print(s1[(s1>35) | (s1<50)])# double condition

```

```python
### Q56. Detect outliers in a column.
#> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.
Q1 = df["emp_salary"].quantile(0.25)
Q3 = df["emp_salary"].quantile(0.75)

IQR = Q3 - Q1

r1 = Q1 - 1.5 * IQR
print(r1)
r2 = Q3 + 1.5 * IQR
print(r2)


```

```python
### Q57. Handle outliers using clipping.
#> 💡 **Hint:** `df["col"].clip(lower=lo, upper=hi)` replaces values outside the bounds with the boundary values.
Q1 = df["emp_salary"].quantile(0.25)
Q3 = df["emp_salary"].quantile(0.75)
IQR = Q3 - Q1

lo = Q1 - 1.5 * IQR
hi = Q3 + 1.5 * IQR
df["emp_salary_clipped"] = df["emp_salary"].clip(lower=lo, upper=hi)
print(df)
```

```python
import numpy as np
import pandas as pd

# 1. Load data
Data = {
    "Name": ["A", "B", "C", "D"],
    "Marks": [89, np.nan, 98, np.nan]
}

df = pd.DataFrame(Data)

# 2. Inspect data
print("shape:", df.shape)
print("dtypes:\n", df.dtypes)
print("nulls:\n", df.isnull().sum())

# 3. Clean data
df = df.fillna(df["Marks"].mean())   # better than 0
df = df.drop_duplicates()

# 4. Feature Engineering
df["Marks_Square"] = np.square(df["Marks"])
df["Marks_Log"] = np.log1p(df["Marks"])

# 5. GroupBy / Pivot
pivot = pd.pivot_table(
    df,
    values="Marks",
    index="Name",
    aggfunc="mean"
)

print("\nPivot Table:\n", pivot)

# 6. Rank + Sort
df["Rank"] = df["Marks"].rank(ascending=False, method="min")
df = df.sort_values("Rank")

print("\nFinal Data:\n", df)

# 7. Export
df.to_csv("pandas_pipeline.csv", index=False)
```
